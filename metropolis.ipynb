```python
import numpy as np
import matplotlib.pyplot as plt


# --------------------------------------------------
# Target density: Cauchy distribution
# In Metropolis, only a proportional form is enough
# --------------------------------------------------
def cauchy(theta):
    return 1 / (1 + theta**2)


# --------------------------------------------------
# Initiation of Metropolis algorithm
# --------------------------------------------------
T = 5000                 # Maximum number of iterations
sigma = 1               # Standard deviation of Normal proposal density
theta_min = -30
theta_max = 30          # Define a range for theta

theta = np.zeros(T)     # Storage space for the samples

np.random.seed(0)       # Similar to rng('default') in MATLAB

theta[0] = np.random.uniform(theta_min, theta_max)   # Initial value


# --------------------------------------------------
# Start sampling
# --------------------------------------------------
t = 0
while t < T - 1:
    t = t + 1

    # Generate proposal from Normal(theta[t-1], sigma)
    theta_star = np.random.normal(theta[t - 1], sigma)

    # Acceptance probability
    alpha = min(1, cauchy(theta_star) / cauchy(theta[t - 1]))

    # Draw a uniform random number
    u = np.random.rand()

    # Accept or reject
    if u < alpha:
        theta[t] = theta_star
    else:
        theta[t] = theta[t - 1]


# --------------------------------------------------
# Plot
# --------------------------------------------------
plt.figure(figsize=(20, 15))

# Display histogram of samples
plt.subplot(3, 1, 1)
nbins = 200
thetabins = np.linspace(theta_min, theta_max, nbins)

counts, bin_edges = np.histogram(theta, bins=thetabins)
bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])

plt.bar(bin_centers, counts / np.sum(counts), width=(bin_edges[1] - bin_edges[0]))
plt.xlim([theta_min, theta_max])
plt.xlabel('theta')
plt.ylabel('p(theta)')

# Theoretical density
y = cauchy(bin_centers)
plt.plot(bin_centers, y / np.sum(y), 'r', linewidth=3)

# Display the history of samples
plt.subplot(3, 1, (2, 3))
plt.plot(theta, np.arange(1, T + 1), 'k')
plt.xlabel('theta')
plt.ylabel('t')
plt.xlim([theta_min, theta_max])
plt.gca().invert_yaxis()   # Similar to set(gca,'yDir','reverse') in MATLAB

plt.tight_layout()
plt.show()
```
